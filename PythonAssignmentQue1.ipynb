```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
employees = 100
wage = 150
n = 50
p = 0.00
revenue = 100

sims = 1300
min_sim_profits =float('-inf') 

while min_sim_profits <= 0:
    sim_conversions_up = [np.sum(np.random.binomial(n, p, size=employees)) for i in range(sims)]
    sim_profits = np.array(sim_conversions_up)*revenue - employees*wage
    p = p+0.001
    min_sim_profits = min(sim_profits)

fig, ax = plt.subplots(figsize=(14,7))
ax = sns.histplot(sim_profits, bins=20, label='No Profit call center simulation results', color='red')
ax.set_xlabel("Profits",fontsize=16)
ax.set_ylabel("Frequency",fontsize=16)
plt.legend()
plt.show()
print("Zero Profit Conversion percentage: ",p)


while min_sim_profits <= 10000:
    sim_conversions_up = [np.sum(np.random.binomial(n, p, size=employees)) for i in range(sims)]
    sim_profits = np.array(sim_conversions_up)*revenue - employees*wage
    p = p+0.001
    min_sim_profits = min(sim_profits)

fig, ax = plt.subplots(figsize=(14,7))
ax = sns.histplot(sim_profits, bins=20, label='Profit of 10000 call center simulation results', color='red')
ax.set_xlabel("Profits",fontsize=16)
ax.set_ylabel("Frequency",fontsize=16)
plt.legend()
plt.show()
print("10000 Profit Conversion percentage: ",p)
```
